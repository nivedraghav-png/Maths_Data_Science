```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('winequality-red.csv', sep=';')

print(df.head())

print(df.info())
```

# 1. Check for missing Values

```python
print(df.isnull().sum())

df.fillna(df.mean(numeric_only=True), inplace=True)
print(df.isnull().sum())
```

# 2: Extract the following columns as vectors: Alcohol and Citric Acid

```python

alcohol = df['alcohol'].to_numpy()
citric_acid = df['citric acid'].to_numpy()

print("Alcohol Vector:")
print(alcohol)

print("Citric Acid Vector:")
print(citric_acid)
```

# 3. Calculate the covariance matrix using np.cov(X.T).

```python
X = df[['alcohol', 'density']].values

cov_matrix = np.cov(X.T) 
print(cov_matrix)
```

# 4. Perform eigen decomposition and identify the top two eigenvalues and corresponding eigenvectors.

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

ev = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[ev]
eigenvectors = eigenvectors[:, ev]

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)
```

# 5. Which wine quality is most common in the dataset? Interpret the distribution.

```python
quality_counts = df['quality'].value_counts()
print(quality_counts)

most_common = df['quality'].mode()[0]
print("Most common wine quality:", most_common)
```

# Interpretation

* Wine quality score 5 is the most common score in the dataset, with 681 samples (approximately 42.59%).

* Quality score 6 is the second most common, with 638 samples (approximately 39.90%).

* Most wines have quality scores between 5 and 6.

* Very few wines received quality scores of 3 or 8.

This indicates that the dataset is concentrated around average-quality wines, with relatively few wines receiving very low or very high quality ratings.
